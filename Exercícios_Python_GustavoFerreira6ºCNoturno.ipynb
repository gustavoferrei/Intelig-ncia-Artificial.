22. Criptografia de Cifra de César

```python
def cifra_cesar(mensagem, deslocamento):
    resultado = ""

    for caractere in mensagem:
        if caractere.isalpha():
            # Define o código base (maiúscula ou minúscula)
            base = ord('A') if caractere.isupper() else ord('a')

            # Aplica o deslocamento com wrap-around usando módulo 26
            novo_char = chr((ord(caractere) - base + deslocamento) % 26 + base)
            resultado += novo_char
        else:
            # Mantém espaços, pontuação e números inalterados
            resultado += caractere

    return resultado
# Exemplo
mensagem = "Bom jogo"
deslocamento = 4
criptografada = cifra_cesar(mensagem, deslocamento)
print(f"Original: {mensagem}")
print(f"Criptografada: {criptografada}")
```

23. Descriptografar Cifra de César

```python
def decifrar_cesar(mensagem, deslocamento):
    return cifra_cesar(mensagem, -deslocamento)

# Exemplo
original = decifrar_cesar("Fsq nsks", 4)
print(original)
```

24. Validação de CPF

```python
def validar_cpf(cpf):
    # Remove caracteres
    cpf = ''.join(filter(str.isdigit, cpf))
    # Verifica se tem 11 dígitos
    if len(cpf) != 11:
        return False
    if cpf == cpf[0] * 11:
        return False
    # Verifica o primeiro dígito
    soma = 0
    for i in range(9):
        soma += int(cpf[i]) * (10 - i)
    resto = soma % 11
    primeiro_digito = 0 if resto < 2 else 11 - resto
    if int(cpf[9]) != primeiro_digito:
        return False
    return True
print(validar_cpf(input("Digite o CPF: ")))
```

30. Jogo da Forca

```python
import random
palavras = ["aurea", "futebol", "internet", "loucura", "jovens"]
p = random.choice(palavras)
ld = ["_"] * len(p) # letras descobertas
tentativas = 6
lu = [] # letras usadas
print("JOGO DA FORCA")
while tentativas > 0 and "_" in ld:
    print("\nPalavra:", " ".join(ld))
    print("Tentativas restantes:", tentativas)
    print("Letras usadas:", ", ".join(lu))
    letra = input("Digite uma letra: ").lower()
    if len(letra) != 1 or not letra.isalpha():
        print("Digite apenas uma letra!")
        continue
    if letra in lu:
        print("Você já tentou essa letra!")
        continue
    lu.append(letra)
    if letra in p:
        print("Acertou!")
        for i in range(len(p)):
            if p[i] == letra:
                ld[i] = letra
    else:
        print("Errou!")
        tentativas -= 1

if "_" not in ld:
    print("Voce descobriu a palavra:", p)
else:
    print("Voce perdeu!")
    print("A palavra era:", p)
```

32. Jogo da Velha

```python
tabuleiro = [" "] * 9
def mostrar_tabuleiro():
    print()
    print(f" {tabuleiro[0]} | {tabuleiro[1]} | {tabuleiro[2]} ")
    print("---+---+---")
    print(f" {tabuleiro[3]} | {tabuleiro[4]} | {tabuleiro[5]} ")
    print("---+---+---")
    print(f" {tabuleiro[6]} | {tabuleiro[7]} | {tabuleiro[8]} ")
    print()
def verificar_vencedor(jogador):
    combinacoes = [
        [0, 1, 2],
        [3, 4, 5],
        [6, 7, 8],
        [0, 3, 6],
        [1, 4, 7],
        [2, 5, 8],
        [0, 4, 8],
        [2, 4, 6]
    ]

    for combinacao in combinacoes:
        if all(tabuleiro[posicao] == jogador for posicao in combinacao):
            return True
    return False
jogador_atual = "X"
for rodada in range(9):
    mostrar_tabuleiro()
    print(f"Jogador {jogador_atual}, escolha uma posicao de 1 a 9.")
    while True:
        try:
            posicao = int(input("Digite a posicao: ")) - 1
            if posicao < 0 or posicao > 8:
                print("Escolha uma posicao entre 1 e 9.")
            elif tabuleiro[posicao] != " ":
                print("Essa posicao ja esta ocupada.")
            else:
                break
        except ValueError:
            print("Digite apenas um numero.")
    tabuleiro[posicao] = jogador_atual
    if verificar_vencedor(jogador_atual):
        mostrar_tabuleiro()
        print(f"Jogador {jogador_atual} venceu")
        break
    if jogador_atual == "X":
        jogador_atual = "O"
    else:
        jogador_atual = "X"
else:
    mostrar_tabuleiro()
    print("Empate")
```

40. Lista de Compras

```python
lista_compras = []
while True:
    print("LISTA DE COMPRAS")
    print("1 - Adicionar item")
    print("2 - Remover item")
    print("3 - Listar itens")
    print("4 - Sair")
    op = input("Escolha uma opção: ")
    #Adiciona um item a lista
    if op == "1":
        item = input("Digite o item que deseja adicionar: ")
        lista_compras.append(item)
        print("Item adicionado com sucesso!")
    #Remove item da lista
    elif op == "2":
        item = input("Digite o item que deseja remover: ")
        if item in lista_compras:
            lista_compras.remove(item)
            print("Item removido com sucesso!")
        else:
            print("Item nao encontrado na lista.")
    #Lista itens
    elif op == "3":
        if len(lista_compras) == 0:
            print("A lista de compras está vazia.")
        else:
            print("\nItens da lista:")
            for i, item in enumerate(lista_compras, start=1):
                print(f"{i}. {item}")
    elif op == "4":
        print("Programa encerrado!")
        break
    else:
        print("Opção invalida. Tente novamente.")
```

44. Fibonacci em Nível

```python
def fibonacci(n):
    if n <= 0:
        return []
    elif n == 1:
        return [0]
    s = [0, 1]
    while len(s) < n:
        proximo = s[-1] + s[-2]
        s.append(proximo)
    return s
print(fibonacci(10))
```

48. Ordenação por Inserção

```python
n = [5, 2, 8, 1, 9, 3]
for i in range(1, len(n)):
    chave = n[i]
    j = i - 1
    while j >= 0 and n[j] > chave:
        n[j + 1] = n[j]
        j -= 1
    n[j + 1] = chave
print(n)
```
